Runs OLAP notebooks 01–06 in order. Each one saves its answers as tables in `workspace.olap`. Then open **`07_findings_and_charts`** and click **Run all** to see the charts and findings. Charts only show when a notebook is run interactively, not through `dbutils.notebook.run`.

**Before running:** the ETL pipeline must have run and `40_gold_validate` must have passed. `00_olap_setup` stops if the latest validation has a FAIL.

| Order | Notebook | Business question | OLAP operations |
|---|---|---|---|
| 1 | `01_rollup_concentration` | BQ1 · where loss is concentrated | Roll-up, ranking, Pareto |
| 2 | `02_drilldown_trend` | BQ2 · is loss per event rising? | Roll-up, drill-down, moving average, slice |
| 3 | `03_slice_dice_state` | BQ3 · which states, relative to population | Drill-across, slice, dice |
| 4 | `04_pivot_seasonality` | BQ4 · when in the year | Pivot, unpivot |
| 5 | `05_cube_income` | BQ5 · income and insured loss | Cube, correlation |
| 6 | `06_data_quality_impact` | How much the answers depend on cleaning | Before/after, slice, drill-through |
| 7 | `07_findings_and_charts` | Findings, recommendations, charts | (reads the saved answers) |

```python
from datetime import datetime
OLAP_NOTEBOOKS = ["01_rollup_concentration", "02_drilldown_trend", "03_slice_dice_state",
                  "04_pivot_seasonality", "05_cube_income", "06_data_quality_impact"]
for nb in OLAP_NOTEBOOKS:
    started = datetime.now()
    print(f"▶ {nb} ...", end=" ")
    try:
        dbutils.notebook.run(f"./{nb}", 1800)
    except Exception:
        print("FAILED")
        raise RuntimeError(f"OLAP stopped in {nb}. Open it and click 'Run all' to see the real error.") from None
    print(f"done in {(datetime.now() - started).seconds}s")
print("All OLAP answers saved in the olap schema. Now open 07_findings_and_charts and click Run all.")
```
